# INPUT and OUTPUT - Integrating Customer Details

```python
#objective is to bring in data from df and use it as input for chatgpt
import sys
sys.path.append('/home/tsonm/code/tsonma/lewagon_cust_targeting')
from src.getdata_utils import load_data

X, _ = load_data()

X.head(5)
```

```python
import pandas as pd
import openai

# Example DataFrame
df = X.head(1)
# Convert DataFrame to string (CSV format)
df_str = df.to_csv(index=False)

# Define system prompt (sets the agent's persona)
system_prompt = """
You are a highly experienced and successful bank representative,
specialized in investments, with a proven track record of helping clients achieve
their financial goals while always acting in their best interest.
You are empathetic, trustworthy, and persuasive in your communication.
"""

# Define user prompt (task instruction)
user_prompt = f"""
Here is the customer data:
{df_str}

Here is the definition of each of the columns for additional context:

age → Client’s age (numeric).

job → Type of job (e.g., admin., blue-collar, student, etc.).

marital → Marital status (single, married, divorced, unknown).

education → Education level (basic, high school, university, etc.).

default → Has credit in default? (yes/no/unknown).

balance → Average yearly account balance.

housing → Has housing loan? (yes/no/unknown).

loan → Has personal loan? (yes/no/unknown).

contact → Communication type (cellular, telephone).

day_of_week → Last contact day of the week (mon to fri).

month → Last contact month of year (jan to dec).

campaign → Number of contacts during this campaign (for this client).

pdays → Days since last contact from a previous campaign (negative 1 means never contacted).

previous → Number of contacts performed before this campaign.


Please generate a personalized call script for each customer based on the data provided with a dash of humor
"""

# API request
response = openai.chat.completions.create(
    model="gpt-4.1",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ],
    max_tokens=100
)

# Output
print(response.choices[0].message.content)
```
